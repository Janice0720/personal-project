```python
from transformers import LlamaTokenizer

model_name_or_path = "c:\llama2\chinese-llama-2-7b"
tokenizer = LlamaTokenizer.from_pretrained(model_name_or_path)

sentence = "通义千问-72B（Qwen-72B）是阿里云研发的通义千问大模型系列的720亿参数规模的模型。Qwen-72B是基于Transformer的大语言模型, 在超大规模的预训练数据上进行训练得到。预训练数据类型多样，覆盖广泛，包括大量网络文本、专业书籍、代码等。同时，在Qwen-72B的基础上，我们使用对齐机制打造了基于大语言模型的AI助手Qwen-72B-Chat。本仓库为Qwen-72B的仓库。"
tokenized_sentence = tokenizer.tokenize(sentence)

print(f'length = {len(tokenized_sentence)}')
print(tokenized_sentence)
```

```python
from transformers import AutoTokenizer

model_name_or_path = "C:\llama2\Qwen-1_8B"
tokenizer = AutoTokenizer.from_pretrained(model_name_or_path, trust_remote_code=True)

sentence = "通义千问-72B（Qwen-72B）是阿里云研发的通义千问大模型系列的720亿参数规模的模型。Qwen-72B是基于Transformer的大语言模型, 在超大规模的预训练数据上进行训练得到。预训练数据类型多样，覆盖广泛，包括大量网络文本、专业书籍、代码等。同时，在Qwen-72B的基础上，我们使用对齐机制打造了基于大语言模型的AI助手Qwen-72B-Chat。本仓库为Qwen-72B的仓库。"
tokenized_sentence = tokenizer.tokenize(sentence)
tokens = [item.decode('utf-8') for item in tokenized_sentence]

print(f'length = {len(tokens)}')

for token in tokens:
    print(token, end=',')
```
